# CIR Experiment Analysis

Load experiment outputs and generate paper-ready figures.

Replace `EXPERIMENT_DIR` with the path to a completed run.

```python
import json
import sys
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.insert(0, str(Path.cwd().parent))

sns.set_theme(style="whitegrid")
%matplotlib inline
```

```python
EXPERIMENT_DIR = Path("../outputs/cir_colored_mnist_20250709_120000")

with open(EXPERIMENT_DIR / "metrics.json") as f:
    history = json.load(f)

with open(EXPERIMENT_DIR / "config.json") as f:
    config = json.load(f)

print(f"Method: {config['method']}, Dataset: {config['dataset']}")
print(f"Epochs: {len(history['train_loss'])}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(history["train_loss"], label="Train")
if history.get("eval_loss"):
    axes[0].plot(
        np.linspace(0, len(history["train_loss"]) - 1, len(history["eval_loss"])),
        history["eval_loss"], label="Eval"
    )
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Loss")
axes[0].legend()
axes[0].set_title("Training Curve")

if history.get("itg_drop"):
    axes[1].plot(history["itg_drop"], marker="o")
    axes[1].set_xlabel("Eval Epoch")
    axes[1].set_ylabel("ITG Accuracy Drop (%)")
    axes[1].set_title("Interventional Transfer Gap")

plt.tight_layout()
```

```python
if history.get("acs"):
    plt.figure(figsize=(6, 4))
    plt.plot(history["acs"], marker="s", color="green")
    plt.xlabel("Eval Epoch")
    plt.ylabel("ACS Concentration")
    plt.title("Axis-Wise Causal Sensitivity")
    plt.tight_layout()
```

```python
print("Final metrics:")
if history["train_loss"]:
    print(f"  Train loss: {history['train_loss'][-1]:.4f}")
if history.get("itg_drop"):
    print(f"  ITG drop: {history['itg_drop'][-1]:.2f}%")
if history.get("acc_corr"):
    print(f"  Acc corr: {history['acc_corr'][-1]:.3f}")
if history.get("acc_uncorr"):
    print(f"  Acc uncorr: {history['acc_uncorr'][-1]:.3f}")
if history.get("acs"):
    print(f"  ACS: {history['acs'][-1]:.4f}")
```
